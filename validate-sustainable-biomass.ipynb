```python
import nomenclature
import pyam
```

```python
from nomenclature.processor import DataValidator
```

```python
bioenergy_validator = DataValidator.from_file(
    "validate_data/sustainable_bioenergy.yaml",
    output_path="bioenergy.xlsx"
)
```

```python
filter = bioenergy_validator.criteria_items[0].filter_args
```

```python
df = pyam.read_iiasa("scenariocompass-dev", **filter)
```

```python
df.filter(year=range(2010, 2101)).plot()
```

```python
bioenergy_validator.apply(df)
```

```python
df.filter(year=range(2010, 2101)).plot(color="Reason For Concern|Sustainable Bioenergy Use|World", linewidth=0.3)
```
