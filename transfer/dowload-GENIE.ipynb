```python
import pyam
from pathlib import Path
```

```python
connection = pyam.iiasa.Connection("genie")
```

```python
connection.index().reset_index().scenario.unique()
```

```python
for i, (model, scenario, version) in connection.index().reset_index().iterrows():
    if scenario.startswith("EN_NPi2020-DACl"):  # EN_NPi2020-DACm
        try:
            df = connection.query(model=model, scenario=scenario)
            df.meta.drop(columns="version", inplace=True)
            df.to_excel(f"raw/GENIE/{model}-{scenario}.xlsx")
            print(f"Downloaded {model} - {scenario}")
        except:
            print(f"Failed {model} - {scenario}")
```
