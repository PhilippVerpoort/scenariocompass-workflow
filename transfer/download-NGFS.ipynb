```python
from datetime import datetime
import pandas as pd
from pathlib import Path

import pyam
```

```python
connection = pyam.iiasa.Connection("ngfs_phase_5_internal")
```

```python
scenario_mapping = {
    "o_2c": "Below 2°C",
    "o_1p5c": "Net-Zero 2050",
    "d_delfrag": "Delayed Transition",
    "d_strain": "Fragmented World",
    "h_cpol": "Current Policies",
    "h_ndc": "Nationally Determined Contributions (NDCs)",
    "o_lowdem": "Low Demand",
    "forecast baseline": "Baseline",
}
```

```python
runs = connection.properties(
    default_only=False,
    model=["GCAM 6.0 NGFS", "MESSAGEix-GLOBIOM 2.0-M-R12-NGFS", "REMIND-MAgPIE 3.3-4.8"],
    scenario=scenario_mapping.keys(), 
)
```

```python
runs
```

```python
runs.create_date = pd.to_datetime(runs.create_date, yearfirst=True)
```

```python
release_runs = runs[runs.create_date < datetime(2024, 11, 4)]
```

```python
release_runs
```

```python
for (model, scenario), i in release_runs.groupby(["model", "scenario"]):
    print(model, scenario, max(i.version), max(i.create_date))
```

```python
for (model, scenario), i in release_runs.groupby(["model", "scenario"]):
    print(model, scenario, max(i.version), max(i.create_date))
```

```python
for (model, scenario), i in release_runs.groupby(["model", "scenario"]):
    if model in ["REMIND-MAgPIE 3.3-4.8"]: #["GCAM 6.0 NGFS", "MESSAGEix-GLOBIOM 2.0-M-R12-NGFS", "REMIND-MAgPIE 3.3-4.8"]:
        if scenario in scenario_mapping.keys():
            try:
                version=max(i.version)
                df = connection.query(model=model, scenario=scenario, version=version)
                df.meta.drop(columns="version", inplace=True)
                df.rename(scenario=scenario_mapping, inplace=True)
                df.to_excel(f"raw/NGFS/{model.replace(" ", "_")}-{scenario}-{version}.xlsx")
                print(f"Downloaded {model} - {scenario} - {version}")
            except:
                print(f"Failed {model} - {scenario} - {version}")
```
