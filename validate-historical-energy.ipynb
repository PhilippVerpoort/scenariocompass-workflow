```python
import nomenclature
import pyam
```

```python
from nomenclature.processor import DataValidator
```

```python
energy_validator = DataValidator.from_file(
    "validate_data/historical_reference_IEA.yaml",
    output_path="historical_energy_vetting.xlsx"
)
```

```python
variable_list = list()

for criterion in energy_validator.criteria_items:
    variable_list.extend(criterion.filter_args["variable"])
```

```python
variable_list
```

```python
filter_args = dict(
    variable=list(set(variable_list)),
    region="World",
)
```

```python
df = pyam.read_iiasa("scenariocompass-dev", **filter_args)
```

```python
df_ret = energy_validator.apply(df)
```
